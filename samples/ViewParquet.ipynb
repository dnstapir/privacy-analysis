# View parquet histogram
View and audit downloaded parquet files from the `downloaded_parquets` folder.

```python
import pandas as pd
from pathlib import Path

LOCAL_DIR = Path(".")

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

## View content

```python
files = sorted(LOCAL_DIR.glob("*.parquet"))
print([f.name for f in files])
pdf = pd.read_parquet(files[0])
pdf.head(100)
```

## Audit: verify start_time has no seconds

```python
times = pd.to_datetime(pdf["start_time"])
unique_seconds = times.dt.second.unique()

print(f"File: {files[0].name}")
print(f"Total rows: {len(pdf)}")
print(f"Unique start_time values: {times.nunique()}")
print(f"Unique seconds found: {sorted(unique_seconds)}")

if (unique_seconds == 0).all():
    print("\nVERIFIED: all start_time values have seconds = 0")
else:
    print(f"\nWARNING: found non-zero seconds in {(times.dt.second != 0).sum()} rows")

times.value_counts().sort_index()
```
